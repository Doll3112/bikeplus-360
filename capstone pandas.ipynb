```python
data = pd.merge(orders, order_items, on='order_id')
data
```

```python
recent_date = data['order_date'].max()
recent_date
```

```python
recency = data.groupby('customer_id')['order_date'].max()

recency = pd.DataFrame(recency)

recency['Recency'] = (recent_date - recency['order_date']).dt.days

recency
```

```python
frequency = data.groupby('customer_id')['order_id'].count()

frequency = pd.DataFrame(frequency)

frequency.columns = ['Frequency']

frequency
```

```python
monetary = data.groupby('customer_id')['total_price'].sum()

monetary = pd.DataFrame(monetary)

monetary.columns = ['Monetary']

monetary
```

```python
rfm = pd.merge(recency, frequency, on='customer_id')

rfm = pd.merge(rfm, monetary, on='customer_id')

rfm
```

```python
rfm = rfm.drop('order_date', axis=1)

rfm
```

```python
def segment(x):
    if x < 1000:
        return "Low"
    elif x < 5000:
        return "Medium"
    else:
        return "High"
```

```python
rfm['Segment'] = rfm['Monetary'].apply(segment)

rfm
```

```python
from sqlalchemy import create_engine

engine = create_engine("mysql+pymysql://root:12345@localhost:3306/retail_sales")
```

```python
rfm.to_sql(
    'customer_segments',
    engine,
    if_exists='replace',
    index=True
)
```

```python
pd.read_sql("SELECT * FROM customer_segments", engine)
```
